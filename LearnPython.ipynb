```python
cart_items = []
cart_prices = []


# Function to validate item name
def get_valid_item_name():

    while True:

        item_name = input("Enter item name: ")

        # Remove front and back spaces
        item_name = item_name.strip()

        # Check if item name is empty
        if item_name == "":
            print("Item name cannot be empty.")

        # Check if item name contains only alphabets and spaces
        elif item_name.replace(" ", "").isalpha():
            return item_name

        else:
            print("Invalid item name. Use letters only.")


# Function to get valid item price
def get_valid_price():

    while True:

        price = float(input("Enter item price: "))

        if price > 0:
            return price

        else:
            print("Price must be greater than 0.")


# Function to add items into cart
def add_items():

    number_of_items = int(input("How many items do you want to add? "))

    for i in range(number_of_items):

        print("\nItem", i + 1)

        item_name = get_valid_item_name()
        item_price = get_valid_price()

        cart_items.append(item_name)
        cart_prices.append(item_price)


# Function to display cart items
def show_cart():

    print("\nItems in Cart:")

    for i in range(len(cart_items)):
        print(cart_items[i], "-", cart_prices[i])


# Function to calculate bill
def calculate_bill():

    total_cost = sum(cart_prices)

    print("\nTotal Cost =", total_cost)

    discount = 0

    apply_discount = input(
        "Do you want to apply discount? (yes/no): "
    )

    if apply_discount.lower() == "yes":

        minimum_purchase = float(
            input("Enter minimum purchase amount for discount: ")
        )

        discount_percentage = float(
            input("Enter discount percentage: ")
        )

        if total_cost >= minimum_purchase:

            discount = total_cost * (discount_percentage / 100)

            print(
                discount_percentage,
                "% discount applied!"
            )

        else:
            print("Cart value is below minimum purchase amount.")

    final_amount = total_cost - discount

    print("Discount Amount =", discount)
    print("Final Amount =", final_amount)


# Main Program
print("Welcome to Shopping Cart System")

add_items()

show_cart()

calculate_bill()

print("\nThank you for shopping!")
```

```python
# Dictionary to store catalog
catalog = {}

# Shopping cart
cart = {}


# Function to validate item name
def get_valid_item_name():

    while True:

        item_name = input("Enter item name: ")

        item_name = item_name.strip().lower()

        if item_name == "":
            print("Item name cannot be empty.")

        elif item_name.replace(" ", "").isalpha():
            return item_name

        else:
            print("Invalid item name.")
            print("Only letters and spaces allowed.")


# Function to validate measure type
def get_valid_measure():

    valid_measures = [
        "kg",
        "quantity",
        "dozen",
        "liters"
    ]

    while True:

        measure = input(
            "Enter measure (kg/quantity/dozen/liters): "
        )

        measure = measure.strip().lower()

        if measure in valid_measures:
            return measure

        else:
            print("Invalid measure type.")


# Function to validate price
def get_valid_price():

    while True:

        price = float(input("Enter item price: "))

        if price > 0:
            return price

        else:
            print("Price must be greater than 0.")


# Function to add item into catalog
def add_item_to_catalog():

    item_name = get_valid_item_name()

    if item_name in catalog:

        print("Item already exists.")

    else:

        measure = get_valid_measure()

        price = get_valid_price()

        # Store item details
        catalog[item_name] = {
            "measure": measure,
            "price": price
        }

        print("Item added successfully!")


# Function to show catalog
def show_catalog():

    print("\nAvailable Items")

    for item, details in catalog.items():

        print(
            item,
            "-",
            details["price"],
            "per",
            details["measure"]
        )


# Function to add items into cart
def add_to_cart():

    number_of_items = int(
        input("\nHow many items do you want to buy? ")
    )

    for i in range(number_of_items):

        print("\nItem", i + 1)

        item_name = input(
            "Enter item name to purchase: "
        )

        item_name = item_name.strip().lower()

        if item_name in catalog:

            quantity = float(
                input(
                    "Enter quantity in "
                    + catalog[item_name]["measure"]
                    + ": "
                )
            )

            cart[item_name] = quantity

            print("Item added to cart.")

        else:
            print("Item not available in catalog.")


# Function to calculate bill
def calculate_bill():

    print("\nFinal Bill")

    total_cost = 0

    for item, quantity in cart.items():

        item_price = catalog[item]["price"]

        item_total = quantity * item_price

        total_cost = total_cost + item_total

        print(
            item,
            "-",
            quantity,
            catalog[item]["measure"],
            "x",
            item_price,
            "=",
            item_total
        )

    print("\nTotal Cost =", total_cost)

    discount = 0

    apply_discount = input(
        "\nApply discount? (yes/no): "
    )

    if apply_discount.lower() == "yes":

        minimum_purchase = float(
            input(
                "Enter minimum purchase amount: "
            )
        )

        discount_percentage = float(
            input(
                "Enter discount percentage: "
            )
        )

        if total_cost >= minimum_purchase:

            discount = (
                total_cost
                * discount_percentage
                / 100
            )

            print(
                discount_percentage,
                "% discount applied."
            )

        else:
            print("Discount not applicable.")

    final_amount = total_cost - discount

    print("Discount Amount =", discount)

    print("Final Amount =", final_amount)


# MAIN PROGRAM

print("WELCOME TO SHOP MANAGEMENT SYSTEM")

# Shopkeeper adds catalog items
number_of_catalog_items = int(
    input(
        "\nHow many catalog items do you want to add? "
    )
)

for i in range(number_of_catalog_items):

    print("\nAdding Catalog Item", i + 1)

    add_item_to_catalog()


# Show catalog to customer
show_catalog()


# Customer shopping
add_to_cart()


# Generate final bill
calculate_bill()


print("\nThank you for shopping!")
```

```python
print("=== String Extractor Program ===")

# ✅ Step 1: Take string input (no restriction)
while True:
    user_input = input("Enter a string: ").strip()

    # Only check empty input
    if user_input == "":
        print("❌ Input cannot be empty.")
    else:
        break


# ✅ Step 2: Get start index
while True:
    try:
        start = int(input("Enter start index: "))
        if start < 0:
            print("❌ Start index cannot be negative.")
            continue
        break
    except ValueError:
        print("❌ Please enter a valid integer.")


# ✅ Step 3: Get end index
while True:
    try:
        end = int(input("Enter end index: "))
        if end < 0:
            print("❌ End index cannot be negative.")
            continue
        if end < start:
            print("❌ End index must be ≥ start index.")
            continue
        break
    except ValueError:
        print("❌ Please enter a valid integer.")


# ✅ Step 4: Extract substring safely
length = len(user_input)

if start >= length:
    print("❌ Start index is out of range.")
else:
    extracted = user_input[start:end]

    print("\nOriginal String:", user_input)
    print("Extracted Substring:", extracted)
```

```python
print("=== Advanced String Extractor (with Step) ===")

# ✅ Step 1: Take string input
while True:
    user_input = input("Enter a string: ").strip()
    if user_input == "":
        print("❌ Input cannot be empty.")
    else:
        break


# ✅ Step 2: Start index
while True:
    try:
        start = int(input("Enter start index: "))
        break
    except ValueError:
        print("❌ Enter a valid integer.")


# ✅ Step 3: End index
while True:
    try:
        end = int(input("Enter end index: "))
        break
    except ValueError:
        print("❌ Enter a valid integer.")


# ✅ Step 4: Step (increment)
while True:
    try:
        step = int(input("Enter step value: "))
        if step == 0:
            print("❌ Step cannot be zero.")
            continue
        break
    except ValueError:
        print("❌ Enter a valid integer.")


# ✅ Step 5: Extract substring using slicing
try:
    extracted = user_input[start:end:step]

    print("\nOriginal String:", user_input)
    print("Extracted Substring:", extracted)

except Exception as e:
    print("❌ Error during extraction:", e)
```

```python
print("=== Student Management System ===")

students = {}   # ✅ Dictionary to store student data


while True:
    print("\nMenu:")
    print("1. Add Student")
    print("2. Retrieve Student")
    print("3. Display All Students")
    print("4. Calculate Average Grade")
    print("5. Exit")

    choice = input("Enter your choice: ").strip()


    # ✅ 1. Add Student
    if choice == "1":

        # ✅ Name validation (your logic integrated)
        while True:
            name = input("Enter student name: ").strip()

            if name == "":
                print("❌ Name cannot be empty.")
                continue

            if not name.replace(" ", "").isalpha():
                print("❌ Invalid name. Only letters and spaces allowed.")
                continue

            break


        # ✅ Prevent duplicate student
        if name in students:
            print("❌ Student already exists.")
            continue


        # ✅ Age validation
        while True:
            try:
                age = int(input("Enter age: "))
                if age > 0:
                    break
                else:
                    print("❌ Age must be positive.")
            except ValueError:
                print("❌ Invalid age.")


        # ✅ Grades input
        while True:
            grades_input = input("Enter grades separated by space: ").strip()

            try:
                grades = []
                for g in grades_input.split():
                    grades.append(float(g))
                break
            except ValueError:
                print("❌ Enter valid numeric grades.")


        # ✅ Store data
        students[name] = {
            "age": age,
            "grades": grades
        }

        print(f"✅ Student '{name}' added successfully.")


    # ✅ 2. Retrieve Student
    elif choice == "2":
        name = input("Enter student name: ").strip()

        if name in students:
            print("\nStudent Details:")
            print("Name:", name)
            print("Age:", students[name]["age"])
            print("Grades:", students[name]["grades"])
        else:
            print("❌ Student not found.")


    # ✅ 3. Display All Students
    elif choice == "3":
        if not students:
            print("❌ No students available.")
        else:
            print("\nAll Students:")
            for name, details in students.items():
                print(f"{name} → Age: {details['age']}, Grades: {details['grades']}")


    # ✅ 4. Calculate Average Grade
    elif choice == "4":
        name = input("Enter student name: ").strip()

        if name in students:
            grades = students[name]["grades"]

            if len(grades) == 0:
                print("❌ No grades available.")
            else:
                avg = sum(grades) / len(grades)
                avg = round(avg, 2)

                if avg.is_integer():
                    avg = int(avg)

                print(f"Average grade of {name} = {avg}")
        else:
            print("❌ Student not found.")


    # ✅ 5. Exit
    elif choice == "5":
        print("Exiting program...")
        break

    else:
        print("❌ Invalid choice.")
```

```python
print("=== Set Intersection Program ===")

# ✅ Step 1: Take Set 1 input
while True:
    input1 = input("Enter elements of Set 1 (space-separated): ").strip()

    if input1 == "":
        print("❌ Input cannot be empty.")
        continue

    try:
        set1 = set()

        for item in input1.split():
            # ✅ Convert to int OR float
            if "." in item:
                number = float(item)
            else:
                number = int(item)

            set1.add(number)

        break

    except ValueError:
        print("❌ Only numeric values are allowed.")


# ✅ Step 2: Take Set 2 input
while True:
    input2 = input("Enter elements of Set 2 (space-separated): ").strip()

    if input2 == "":
        print("❌ Input cannot be empty.")
        continue

    try:
        set2 = set()

        for item in input2.split():
            if "." in item:
                number = float(item)
            else:
                number = int(item)

            set2.add(number)

        break

    except ValueError:
        print("❌ Only numeric values are allowed.")


# ✅ Step 3: Calculate intersection
intersection = set1.intersection(set2)


# ✅ Step 4: Display result
print("\nSet 1:", set1)
print("Set 2:", set2)

if len(intersection) == 0:
    print("❌ No common elements found between the two sets.")
else:
    print("✅ Intersection:", intersection)
```

```python
print("=== Inventory Management System ===")

inventory = {}   # Dictionary to store items

while True:
    print("\nMenu:")
    print("1. Add Item")
    print("2. Update Item")
    print("3. Display Inventory")
    print("4. Exit")

    choice = input("Enter your choice: ").strip()


    # ✅ 1. ADD ITEM
    if choice == "1":

        # ✅ Item name validation
        while True:
            item_name = input("Enter item name: ").strip()

            if item_name == "":
                print("❌ Item name cannot be empty.")
                continue

            if not item_name.replace(" ", "").isalpha():
                print("❌ Item name must contain only letters and spaces.")
                continue

            if item_name in inventory:
                print("❌ Item already exists.")
                continue

            break

        # ✅ Quantity validation
        while True:
            qty_input = input("Enter quantity: ").strip()

            try:
                quantity = int(qty_input)

                if quantity <= 0:
                    print("❌ Quantity must be greater than 0.")
                    continue

                break

            except ValueError:
                print("❌ Quantity must be a number.")

        inventory[item_name] = quantity

        print("✅ Item added successfully.")


    # ✅ 2. UPDATE ITEM
    elif choice == "2":
        item_name = input("Enter item name to update: ").strip()

        if item_name in inventory:

            while True:
                qty_input = input("Enter new quantity: ").strip()

                try:
                    quantity = int(qty_input)

                    if quantity <= 0:
                        print("❌ Quantity must be greater than 0.")
                        continue

                    break

                except ValueError:
                    print("❌ Quantity must be a number.")

            inventory[item_name] = quantity

            print("✅ Item quantity updated successfully.")

        else:
            print("❌ Item not found.")


    # ✅ 3. DISPLAY INVENTORY
    elif choice == "3":
        if not inventory:
            print("❌ Inventory is empty.")
        else:
            print("\nInventory:")
            for item, qty in inventory.items():
                print(f"{item} → {qty}")


    # ✅ 4. EXIT
    elif choice == "4":
        print("✅ Exiting Inventory System.")
        break

    else:
        print("❌ Invalid choice. Please try again.")
```

```python
print("=== Unique Words Program ===")

# ✅ Step 1: Strict input validation
while True:
    sentence = input("Enter a sentence: ").strip()

    # ❌ Empty check
    if sentence == "":
        print("❌ Input cannot be empty.")
        continue

    # ❌ Allow ONLY letters and spaces
    valid = True
    for char in sentence:
        if not (char.isalpha() or char.isspace()):
            valid = False
            break

    if not valid:
        print("❌ Only letters and spaces are allowed. Enter a valid sentence.")
        continue

    break


# ✅ Step 2: Split into words
words = sentence.split()

# ✅ Step 3: Normalize (lowercase)
normalized_words = []
for w in words:
    normalized_words.append(w.lower())

# ✅ Step 4: Extract unique words
unique_words = set(normalized_words)


# ✅ Output
print("\n--- Results ---")
print("Words:", words)
print("Unique Words:", unique_words)
```

```python
Write a Python program using functions and Object-Oriented Programming (OOP) concepts to manage student details and calculate their average grades.

---------------------------------------
FUNCTIONAL REQUIREMENTS
---------------------------------------

1. Accept Student Details:
   - Create a class named "Student"
   - Each student should have:
       - Name (string)
       - Age (integer)
       - Grades (list of numbers)
   - Allow the user to input multiple students

2. Calculate Average Grade:
   - Implement a method inside the Student class to calculate the average of grades
   - Display the average grade for each student

3. Display Student Summary:
   - Display all student details in a structured format:
       Name → Age → Grades → Average Grade

---------------------------------------
VALIDATION REQUIREMENTS
---------------------------------------

- Name:
    - Must not be empty
    - Must contain only letters and spaces

- Age:
    - Must be a positive integer

- Grades:
    - Must be numeric (integer or float)
    - At least one grade must be entered
    - Reject invalid values (letters, special characters)

- Handle invalid inputs gracefully using try-except
- Ensure program does not crash

---------------------------------------
ADDITIONAL REQUIREMENTS
---------------------------------------

- Use:
    - Classes (OOP)
    - Methods inside class
    - Functions (to organize logic where appropriate)

- Use a menu-driven approach:
    1. Add Student
    2. Display All Students
    3. Exit

- Allow multiple students to be added
- Display clear and user-friendly messages

---------------------------------------
EXAMPLE INPUT/OUTPUT
---------------------------------------

Menu:
1. Add Student
2. Display Students
3. Exit

Enter choice: 1

Enter name: John Doe
Enter age: 18
Enter grades (space-separated): 80 90 85

✅ Student added successfully

Enter choice: 2

Students:
John Doe → Age: 18 → Grades: [80, 90, 85] → Average: 85.0

---------------------------------------

---------------------------------------
ADVANCED (OPTIONAL)
---------------------------------------

- Store student objects in a list
- Sort students by average grade
- Handle duplicate student names
```

```python
print("=== Student Management System ===")

# ✅ Class Definition
class Student:
    def __init__(self, name, age, grades):
        self.name = name
        self.age = age
        self.grades = grades

    def calculate_average(self):
        if len(self.grades) == 0:
            return 0
        avg = sum(self.grades) / len(self.grades)

        # ✅ Clean output (int if whole number)
        if avg.is_integer():
            return int(avg)
        return round(avg, 2)


# ✅ Function to validate name
def get_valid_name():
    while True:
        name = input("Enter name: ").strip()

        if name == "":
            print("❌ Name cannot be empty.")
            continue

        if not name.replace(" ", "").isalpha():
            print("❌ Only letters and spaces allowed.")
            continue

        return name


# ✅ Function to validate age
def get_valid_age():
    while True:
        try:
            age_input = input("Enter age: ").strip()

            # ❌ Empty check
            if age_input == "":
                print("❌ Age cannot be empty.")
                continue

            age = int(age_input)

            # ❌ Lower bound
            if age <= 0:
                print("❌ Age must be greater than 0.")
                continue

            # ✅ Upper bound (realistic human range)
            if age > 120:
                print("❌ Age must be realistic (<= 120).")
                continue
           
            # student-age specific validation
            if age < 3 or age > 30:
                print("❌ Age must be between 3 and 30 (student range).")
                continue

                
            return age

        except ValueError:
            print("❌ Age must be a valid integer.")


# ✅ Function to get grades
def get_valid_grades():
    while True:
        grades_input = input("Enter grades (space-separated): ").strip()

        if grades_input == "":
            print("❌ At least one grade required.")
            continue

        try:
            grades = []
            for g in grades_input.split():
                value = float(g)
                grades.append(value)

            return grades

        except ValueError:
            print("❌ Invalid grades. Enter numeric values only.")


# ✅ Function to display students
def display_students(student_list):
    if not student_list:
        print("❌ No students available.")
        return

    print("\nStudents:")
    for student in student_list:
        avg = student.calculate_average()
        print(f"{student.name} → Age: {student.age} → Grades: {student.grades} → Average: {avg}")


# ✅ MAIN PROGRAM
students = []

while True:
    print("\nMenu:")
    print("1. Add Student")
    print("2. Display All Students")
    print("3. Exit")

    choice = input("Enter choice: ").strip()


    # ✅ Add Student
    if choice == "1":
        name = get_valid_name()

        # ✅ Prevent duplicate names
        if any(s.name.lower() == name.lower() for s in students):
            print("❌ Student already exists.")
            continue

        age = get_valid_age()
        grades = get_valid_grades()

        student = Student(name, age, grades)
        students.append(student)

        print("✅ Student added successfully.")


    # ✅ Display Students
    elif choice == "2":
        display_students(students)


    # ✅ Exit
    elif choice == "3":
        print("✅ Exiting program.")
        break


    else:
        print("❌ Invalid choice.")
```

```python
print("=== Basic Calculator ===")

# ✅ Functions for operations
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "❌ Error: Division by zero is not allowed"
    return a / b


# ✅ Function to get valid number (int or float)
def get_number(prompt):
    while True:
        value = input(prompt).strip()

        try:
            if "." in value:
                return float(value)
            else:
                return int(value)
        except ValueError:
            print("❌ Invalid input. Enter a valid number.")


# ✅ Function to get valid operator
def get_operator():
    valid_ops = ["+", "-", "*", "/"]

    while True:
        op = input("Enter operator (+, -, *, /): ").strip()
        if op in valid_ops:
            return op
        else:
            print("❌ Invalid operator.")


# ✅ Main loop
while True:
    print("\n--- New Calculation ---")

    num1 = get_number("Enter first number: ")
    num2 = get_number("Enter second number: ")
    operator = get_operator()

    # ✅ Call appropriate function
    if operator == "+":
        result = add(num1, num2)
    elif operator == "-":
        result = subtract(num1, num2)
    elif operator == "*":
        result = multiply(num1, num2)
    elif operator == "/":
        result = divide(num1, num2)

    # ✅ Result formatting
    if isinstance(result, (int, float)):
        if isinstance(result, float):
            result = round(result, 2)

            # Convert float like 2.0 → 2
            if result.is_integer():
                result = int(result)

    print(f"Result: {result}")

    # ✅ Continue or exit
    choice = input("\nDo you want to continue? (y/n): ").strip().lower()
    if choice != "y":
        print("✅ Exiting Calculator.")
        break
```

```python
print("=== Bookstore Inventory System ===")


# ✅ Class Definition
class Bookstore:
    def __init__(self):
        self.inventory = {}

    # ✅ Add Book
    def add_book(self, title, author, quantity):
        if title in self.inventory:
            self.inventory[title]["quantity"] += quantity
            print("✅ Existing book updated successfully.")
        else:
            self.inventory[title] = {
                "author": author,
                "quantity": quantity
            }
            print("✅ Book added successfully.")

    # ✅ Sell Book
    def sell_book(self, title, quantity):
        if title not in self.inventory:
            print("❌ Book not found.")
        elif self.inventory[title]["quantity"] < quantity:
            print("❌ Not enough stock available.")
        else:
            self.inventory[title]["quantity"] -= quantity
            print("✅ Book sold successfully.")

    # ✅ Display Inventory (UPDATED ✅)
    def display_inventory(self):
        if not self.inventory:
            print("❌ Inventory is empty.")
        else:
            print("\n=== Bookstore Inventory ===")
            for title, details in self.inventory.items():
                print(f"Book Name: {title}")
                print(f"Author: {details['author']}")
                print(f"Quantity: {details['quantity']}")
                print("----------------------------")


# ✅ Validation: Title & Author
def get_valid_text(prompt):
    while True:
        text = input(prompt).strip()

        if text == "":
            print("❌ Value cannot be empty.")
            continue

        if not text.replace(" ", "").isalpha():
            print("❌ Only letters and spaces allowed.")
            continue

        return text


# ✅ Validation: Quantity
def get_valid_quantity(prompt):
    while True:
        try:
            qty_input = input(prompt).strip()

            if qty_input == "":
                print("❌ Quantity cannot be empty.")
                continue

            qty = int(qty_input)

            if qty <= 0:
                print("❌ Quantity must be greater than 0.")
                continue

            return qty

        except ValueError:
            print("❌ Enter a valid number.")


# ✅ MAIN PROGRAM
store = Bookstore()

while True:
    print("\nMenu:")
    print("1. Add Book")
    print("2. Sell Book")
    print("3. Display Inventory")
    print("4. Exit")

    choice = input("Enter choice: ").strip()

    # ✅ Add Book
    if choice == "1":
        title = get_valid_text("Enter title: ")
        author = get_valid_text("Enter author: ")
        quantity = get_valid_quantity("Enter quantity: ")

        store.add_book(title, author, quantity)

    # ✅ Sell Book
    elif choice == "2":
        title = input("Enter title to sell: ").strip()
        quantity = get_valid_quantity("Enter quantity to sell: ")

        store.sell_book(title, quantity)

    # ✅ Display Inventory (FIXED ✅)
    elif choice == "3":
        store.display_inventory()

    # ✅ Exit
    elif choice == "4":
        print("✅ Exiting system.")
        break

    else:
        print("❌ Invalid choice.")
```

```python
print("=== Word Search in File ===")

import string


# ✅ Validate string input (with exit support)
def get_valid_string(prompt):
    while True:
        value = input(prompt).strip()

        # ✅ Exit option
        if value.lower() == "exit":
            print("✅ Program exited.")
            exit()

        if value == "":
            print("❌ Input cannot be empty.")
            continue

        if not any(char.isalpha() for char in value):
            print("❌ Input must contain valid letters.")
            continue

        return value


# ✅ Step 1: Get valid file (with exit option)
while True:
    file_name = input("Enter file name (or type 'exit' to quit): ").strip()

    if file_name.lower() == "exit":
        print("✅ Program exited.")
        exit()

    if file_name == "":
        print("❌ File name cannot be empty.")
        continue

    try:
        with open(file_name, "r") as file:
            lines = file.readlines()
        break

    except FileNotFoundError:
        print("❌ File not found. Try again.")


# ✅ Step 2: Word search loop
while True:
    search_word = get_valid_string("Enter word to search (or type 'exit'): ")
    search_word_lower = search_word.lower()

    total_count = 0
    found = False

    print("\n--- Results ---")

    # ✅ Process file
    for i, line in enumerate(lines, start=1):
        words = line.split()
        count_in_line = 0

        for word in words:
            clean_word = word.strip(string.punctuation).lower()

            if clean_word == search_word_lower:
                count_in_line += 1

        if count_in_line > 0:
            found = True
            total_count += count_in_line

            # ✅ Highlight matches
            temp_line = line
            for word in words:
                if word.strip(string.punctuation).lower() == search_word_lower:
                    temp_line = temp_line.replace(word, f"[{word.upper()}]")

            print(f"Line {i}: {temp_line.strip()}")

    # ✅ Result handling
    if found:
        print(f"\n✅ Total occurrences of '{search_word}': {total_count}")

    else:
        print(f"❌ Word '{search_word}' not found in file.")

    # ✅ Ask to continue or exit
    choice = input("\nDo you want to search another word? (y/n or 'exit'): ").strip().lower()

    if choice in ["n", "exit"]:
        print("✅ Exiting search.")
        break
``
```

```python
import os

print("=== Simple File Copier ===")

# ✅ Step 1: Get valid source file
while True:
    source_file = input("Enter source file name: ").strip()

    if source_file == "":
        print("❌ Source file name cannot be empty.")
        continue

    if not os.path.exists(source_file):
        print("❌ Source file not found. Please try again.")
        continue

    break


# ✅ Step 2: Get destination file
while True:
    dest_file = input("Enter destination file name: ").strip()

    if dest_file == "":
        print("❌ Destination file name cannot be empty.")
        continue

    # ✅ Check if destination already exists
    if os.path.exists(dest_file):
        confirm = input("⚠️ File exists. Overwrite? (y/n): ").lower()

        if confirm != "y":
            print("❌ Enter a different file name.")
            continue

    break


# ✅ Step 3: Copy file contents
try:
    with open(source_file, "r") as src:
        content = src.read()

    with open(dest_file, "w") as dest:
        dest.write(content)

    print("✅ File copied successfully.")

    # ✅ Optional: Show file locations
    print("📂 Source Path:", os.path.abspath(source_file))
    print("📂 Destination Path:", os.path.abspath(dest_file))

except Exception as e:
    print("❌ Error during file copy:", e)
```

```python
print("=== Configuration File Reader ===")

# ✅ Step 1: Get valid file name
while True:
    file_name = input("Enter configuration file name: ").strip()

    if file_name == "":
        print("❌ File name cannot be empty.")
        continue

    try:
        config = {}

        # ✅ Step 2: Read and parse file
        with open(file_name, "r") as file:
            for line in file:
                line = line.strip()

                # ❌ Skip empty lines
                if line == "":
                    continue

                # ❌ Skip invalid format lines
                if "=" not in line:
                    print(f"⚠️ Skipping invalid line: {line}")
                    continue

                # ✅ Split key-value
                key, value = line.split("=", 1)

                # ✅ Normalize
                key = key.strip().lower()
                value = value.strip()

                config[key] = value

        break  # ✅ Exit loop if file read successfully

    except FileNotFoundError:
        print("❌ File not found. Try again.")
    except Exception as e:
        print("❌ Error:", e)


# ✅ Step 3: Display all configurations
print("\n--- Configuration ---")
for key, value in config.items():
    print(f"{key} → {value}")


# ✅ Step 4: Retrieve values loop
while True:
    search_key = input("\nEnter key to search (or type 'exit'): ").strip().lower()

    # ✅ Exit option
    if search_key == "exit":
        print("✅ Exiting program.")
        break

    # ❌ Empty check
    if search_key == "":
        print("❌ Key cannot be empty.")
        continue

    # ✅ Lookup
    if search_key in config:
        print(f"✅ Value: {config[search_key]}")
    else:
        print("❌ Key not found.")
```

```python
✅ ✅ ✅ Prompt: Count Words and Lines in a File (Python)
🧠 Prompt
Write a Python program that reads a file and counts the total number of words and lines present in the file.

---------------------------------------
OBJECTIVE
---------------------------------------
The program should analyze a text file and provide:
1. Total number of words
2. Total number of lines

---------------------------------------
FUNCTIONAL REQUIREMENTS
---------------------------------------

1. Input:
   - Ask the user to enter the file name

2. File Reading:
   - Open the file in read mode
   - Read the file content

3. Counting:
   - Count the number of lines in the file
   - Count the number of words in the file
       - Words can be separated by spaces

4. Output:
   - Display:
       - Total number of lines
       - Total number of words

---------------------------------------
VALIDATION REQUIREMENTS
---------------------------------------

- Ensure the file name is not empty
- Handle FileNotFoundError if the file does not exist
- Handle empty files gracefully (should show 0 counts)
- Ensure program does not crash due to invalid input
- Trim unnecessary spaces before processing

---------------------------------------
FILE HANDLING REQUIREMENTS
---------------------------------------

- Use `with open()` for safe file handling
- Use:
    - `readlines()` for counting lines
    - `split()` for counting words

---------------------------------------
ADDITIONAL REQUIREMENTS
---------------------------------------

- Use a clean and readable output format
- Optionally display file path
- Allow user to retry in case of invalid file name

---------------------------------------
EXAMPLE INPUT/OUTPUT
---------------------------------------

Input:
Enter file name: sample.txt

Output:
Total number of lines: 3
Total number of words: 12

---------------------------------------

Input:
Enter file name: missing.txt

Output:
❌ File not found. Please try again.
```

```python
import os
import string

print("=== Word and Line Counter ===")

# ✅ Step 1: Get valid file name
while True:
    file_name = input("Enter file name: ").strip()

    if file_name == "":
        print("❌ File name cannot be empty.")
        continue

    if not os.path.exists(file_name):
        print("❌ File not found. Please try again.")
        continue

    break


# ✅ Step 2: Read file and count
try:
    with open(file_name, "r") as file:
        lines = file.readlines()

    # ✅ Count lines
    line_count = len(lines)

    # ✅ Count words
    word_count = 0
    for line in lines:
        # words = line.strip().split()   # removes extra spaces
        
        words = [
            word for word in line.split()
            if word.strip(string.punctuation).isalpha()
        ]

        word_count += len(words)

    # ✅ Output
    print("\n--- Results ---")
    print(f"📄 Total number of lines: {line_count}")
    print(f"📝 Total number of words: {word_count}")

    # ✅ Optional: Show file path
    print(f"📂 File location: {os.path.abspath(file_name)}")

except Exception as e:
    print("❌ Error while reading file:", e)
```

```python
✅ ✅ ✅ UPDATED PROMPT (WITH EXIT OPTION)
🧠 Prompt
Write a Python program that takes a file as input and calculates:

1. The total number of words in the file
2. The total number of lines in the file

---------------------------------------
OBJECTIVE
---------------------------------------
The program should read a text file and analyze its content to count:
- Number of lines
- Number of words

---------------------------------------
FUNCTIONAL REQUIREMENTS
---------------------------------------

1. Input:
   - Ask the user to enter the file name
   - Allow the user to type "exit" to terminate the program at any time

2. File Reading:
   - Open the file in read mode
   - Read the file content safely

3. Processing:
   - Count the total number of lines in the file
   - Count the total number of words:
       - Words are separated by spaces

4. Output:
   - Display:
       - Total number of lines
       - Total number of words

---------------------------------------
VALIDATION REQUIREMENTS
---------------------------------------

- Ensure file name is not empty
- Allow user to exit by typing "exit"
- Handle FileNotFoundError if the file does not exist
- If file is not found, prompt user again OR allow exit
- Handle empty file case (should return 0 lines and 0 words)
- Trim extra spaces before counting words
- Ensure the program does not crash due to invalid input

---------------------------------------
FILE HANDLING REQUIREMENTS
---------------------------------------

- Use `with open()` for safe file handling
- Use:
    - `readlines()` method for counting lines
    - `split()` method for counting words

---------------------------------------
ADDITIONAL REQUIREMENTS
---------------------------------------

- Display output in a clean and readable format
- Optionally display the file path
- Provide a retry mechanism for invalid file input
- User should NOT be forced to stay in loop — must have exit option

---------------------------------------
EXAMPLE INPUT/OUTPUT
---------------------------------------

Input:
Enter file name (or type 'exit'): sample.txt

Output:
Total number of lines: 3
Total number of words: 12

---------------------------------------

Input:
Enter file name: missing.txt

Output:
❌ File not found. Try again or type 'exit' to quit.

---------------------------------------

Input:
Enter file name: exit

Output:
✅ Program exited successfully
```

```python
import os
import string
import sys  # ✅ IMPORTANT

print("=== Word and Line Counter (Clean Version) ===")

# ✅ Step 1: Get valid file (with exit option)
while True:
    file_name = input("Enter file name (or type 'exit'): ").strip()

    if file_name.lower() == "exit":
        print("✅ Program exited successfully.")
        sys.exit()   # ✅ FIXED (NOT exit())

    if file_name == "":
        print("❌ File name cannot be empty.")
        continue

    if not os.path.exists(file_name):
        print("❌ File not found. Try again or type 'exit' to quit.")
        continue

    break


# ✅ Step 2: Process file
try:
    with open(file_name, "r") as file:
        lines = file.readlines()

    line_count = 0
    word_count = 0

    for line in lines:
        clean_line = line.strip(string.punctuation + " ").strip()

        if clean_line != "":
            line_count += 1

            words = [
                word.strip(string.punctuation)
                for word in line.split()
                if word.strip(string.punctuation).isalpha()
            ]

            word_count += len(words)

    print("\n--- Results ---")
    print(f"📄 Total number of lines (valid): {line_count}")
    print(f"📝 Total number of words (valid): {word_count}")
    print(f"📂 File location: {os.path.abspath(file_name)}")

except Exception as e:
    print("❌ Error while reading file:", e)
```

```python
import streamlit as st
import pandas as pd
import plotly.express as px

# ✅ Load Data
df = pd.read_excel("Amazon Store Sales Data.xlsx")

# ✅ Prepare data
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Month"] = df["Order Date"].dt.to_period("M").astype(str)

# ✅ Page Title
st.title("📊 Amazon Sales Dashboard")

# =========================
# ✅ KPI SECTION
# =========================
st.subheader("📈 Key Metrics")

col1, col2 = st.columns(2)

col1.metric("Total Sales", f"{df['Sales'].sum():,.0f}")
col2.metric("Total Profit", f"{df['Profit'].sum():,.0f}")

# =========================
# ✅ FILTERS (NEW 🔥)
# =========================
st.sidebar.header("Filters")

category_filter = st.sidebar.multiselect(
    "Select Category",
    options=df["Category"].unique(),
    default=df["Category"].unique()
)

df_filtered = df[df["Category"].isin(category_filter)]

# =========================
# ✅ SALES TREND
# =========================
monthly_sales = df_filtered.groupby("Month")["Sales"].sum().reset_index()

fig1 = px.line(
    monthly_sales,
    x="Month",
    y="Sales",
    title="📈 Sales Trend"
)

st.plotly_chart(fig1)

# =========================
# ✅ CATEGORY SALES
# =========================
category_sales = df_filtered.groupby("Category")["Sales"].sum().reset_index()

fig2 = px.bar(
    category_sales,
    x="Category",
    y="Sales",
    title="📊 Sales by Category"
)

st.plotly_chart(fig2)

# =========================
# ✅ PROFIT vs SALES
# =========================
profit_vs_sales = df_filtered.groupby("Category")[["Sales", "Profit"]].sum().reset_index()

fig3 = px.bar(
    profit_vs_sales,
    x="Category",
    y=["Sales", "Profit"],
    barmode="group",
    title="💸 Profit vs Sales"
)

st.plotly_chart(fig3)

# =========================
# ✅ TOP PRODUCTS
# =========================
top_products = (
    df_filtered.groupby("Product Name")["Sales"]
    .sum()
    .nlargest(10)
    .reset_index()
)

fig4 = px.bar(
    top_products,
    x="Sales",
    y="Product Name",
    orientation="h",
    title="🏆 Top 10 Products"
)

st.plotly_chart(fig4)
```
